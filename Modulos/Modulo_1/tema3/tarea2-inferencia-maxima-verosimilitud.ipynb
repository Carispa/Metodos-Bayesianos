## 1.

Consideramos un conjunto de datos 

$$
\mathcal{D} = \{x_1, \dots, x_N\}
$$

independientes e idénticamente distribuidos, los cuales suponemos que provienen de una distribución normal $\mathcal{N}(\mu, \sigma^2)$.

Muestre que los estimadores de máxima verosimilitud para los parámetros $\mu$ y $\sigma^2$ son:

$$\hat{\mu}_{MLE} = \frac{1}{N} \sum_{j=1}^{N}x_j \qquad \text{y} \qquad \hat{\sigma}_{MLE}^2 = \frac{1}{N}\sum_{j=1}^{N}(x_j-\hat{\mu}_{MLE})^2.$$

Función de Verosimilitud $\mathcal{L}(\mu, \sigma^2)$

$$
\mathcal{L}(\mu, \sigma^2) = p(D, | \mu, \sigma^2) = p(x_1, x_2, \dots, x_n | \mu, \sigma^2)
$$

$$
\mathcal{L}(\mu, \sigma^2) = \frac{1}{\sqrt{2\pi \sigma^2}} e^{-\frac{(x_1 - \mu)^2}{2\sigma^2}} \cdot \dots \cdot \frac{1}{\sqrt{2\pi \sigma^2}} e^{-\frac{(x_n - \mu)^2}{2\sigma^2}}
$$

$$
\mathcal{L}(\mu, \sigma^2) = \left( \frac{1}{\sqrt{2\pi} \sigma} \right)^N e^{-\frac{1}{2\sigma^2} ((x_1 - \mu)^2 + \dots + (x_n - \mu)^2)}
$$

Log-Verosimilitud $l(\mu, \sigma^2)$

$$
l(\mu, \sigma^2) = \log \mathcal{L}(\mu, \sigma^2)
$$

$$
l(\mu, \sigma^2) = N \log \left( \frac{1}{\sqrt{2\pi\sigma^2}} \right) - \frac{1}{2\sigma^2} \sum_{j=1}^{N} (x_i - \mu)^2
$$

Maximizamos respecto a $\mu$ 
$$
\frac{\partial}{\partial \mu} l(\mu, \sigma^2) = 0 \implies - \frac{1}{2\sigma^2} \cdot 2(-1) \sum_{j=1}^{N} (x_j - \mu) = 0
$$
$$
= \frac{1}{\sigma^2} \left( \sum_{j=1}^{N} x_j - \sum_{j=1}^{N} \mu \right) = 0
$$

$$\sum_{j=1}^{N} x_j = N\mu$$

Resultado del estimador de la media
$$\hat{\mu}_{MLE} = \frac{1}{N} \sum_{j=1}^{N} x_j$$

Maximizamos respecto a $\sigma$ 
$$
\frac{\partial}{\partial \sigma} l(\mu, \sigma^2) = 0
$$
$$
= -\frac{N \sqrt{2\pi \sigma^2} \cdot 4\pi \sigma}{2 \cdot 2\pi \sigma^2 \cdot \sqrt{2\pi \sigma^2}} + \frac{1}{\sigma^3} \sum_{i=1}^{N} (x_i - \mu)^2 = 0
$$

$$
= -\frac{N}{\sigma} + \frac{1}{\sigma^3} \sum (x_i - \mu)^2 = 0
$$

Resultado del estimador para la varianza:
$$
\hat{\sigma}^2_{MLE} = \frac{1}{N} \sum_{j=1}^{N} (x_j - \hat{\mu}_{MLE})^2
$$

## 2.

Consideramos un conjunto de datos 

$$
\mathcal{D} = \{x_1, \dots, x_N\}
$$

independientes e idénticamente distribuidos, los cuales suponemos que provienen de una distribución Gamma, con la siguiente pdf

$$
p(x | a) = \frac{a^5}{4!} x^4 e^{-ax}
$$

Encuentre el estimador de máxima verosimilitud para $a$.


Función de Verosimilitud $\mathcal{L}(a)$:

$$\mathcal{L}(a) = p(D|a) = p(x_1, \dots, x_N | a)$$
$$\mathcal{L}(a) = \frac{a^5}{4!} (x_1)^4 e^{-ax_1} \cdot \dots \cdot \frac{a^5}{4!} (x_N)^4 e^{-ax_N}$$
$$\mathcal{L}(a) = \left( \frac{a^5}{4!} \right)^N (x_1)^4 \cdot \dots \cdot (x_N)^4 e^{-a(x_1 + \dots + x_N)}$$

Log-Verosimilitud $l(a)$:

$$
l(a) = \log \mathcal{L}(a) = N \log \left( \frac{a^5}{4!} \right) + 4 \sum_{j=1}^{N} \log(x_j) - a \sum_{j=1}^{N} x_j
$$

Maximizamos:
$$\frac{d}{da} l(a) = 0 \implies N \frac{4!}{a^5} \cdot \frac{5a^4}{4!} - \sum_{j=1}^{N} x_j = 0$$
$$5N \cdot \frac{1}{a} - \sum_{j=1}^{N} x_j = 0$$

Resultado
$$
\hat{a}_{MLE} = 5 \cdot \frac{N}{\sum_{j=1}^{N} x_j}
$$

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from ucimlrepo import fetch_ucirepo 
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
  
# fetch dataset 
abalone = fetch_ucirepo(id=1) 
  
# data (as pandas dataframes) 
X = abalone.data.features 
y = abalone.data.targets 
  
df = pd.concat([X,y], axis=1)
df.head()
```

```python
df.info()
```

```python
# Dividimos el set de datos 
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=19)
```

```python
# Hacemos nuestro pipeline

numeric_cols = x_train.select_dtypes(include=["int64", "float64"]).columns.tolist()
categorical_cols = x_train.select_dtypes(include=["object", "category"]).columns.tolist()

numeric_scaler = Pipeline([
        ("scaler", StandardScaler())
    ])

categorical_encoder = Pipeline([
        ("onehot", OneHotEncoder(drop="first", handle_unknown="ignore"))
    ])

preprocess_scaled = ColumnTransformer([
        ("num", numeric_scaler, numeric_cols),
        ("cat", categorical_encoder, categorical_cols)
    ])

linear_model = Pipeline([
    ("preprocess", preprocess_scaled),
    ("regressor", LinearRegression())
])
```

```python
# Ajustamos el modelo
linear_model.fit(x_train, y_train)
```

```python
# Scores de train y test
train_r = linear_model.score(x_train, y_train)
test_r = linear_model.score(x_test, y_test)

print(f'El score del train es: {train_r}')
print(f'El score del test es: {test_r}')
```

Ahora vamos a hacerlo con regularizacion y vamos a ver cual es el mejor hiperparametro

```python
search = GridSearchCV(Ridge(), {"alpha": np.arange(0.2,10.2,0.2)}, cv=5, scoring="neg_root_mean_squared_error")
search.fit(preprocess_scaled.fit_transform(x_train), y_train)
search.best_estimator_
```

```python
# Ahora para un modelo de regularizacion 
ridge_model = Pipeline([
    ("preprocess", preprocess_scaled),
    ("regressor", Ridge(alpha=2))
])
```

```python
# Ajustamos el modelo
ridge_model.fit(x_train, y_train)
```

```python
# Scores de train y test
trainr_r = ridge_model.score(x_train, y_train)
testr_r = ridge_model.score(x_test, y_test)

print(f'El score del train es: {trainr_r}')
print(f'El score del test es: {testr_r}')
```
