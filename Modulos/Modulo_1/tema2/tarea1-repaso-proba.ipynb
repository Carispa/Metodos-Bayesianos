## 1. 

Supongamos que una prueba para la COVID-19 tiene una precisión del 95%. Esto es, si el paciente está infectado, entonces el resultado de la prueba será positivo con probabilidad de 0.95, y si el paciente NO está infectado, el resutado de la prueba será negativo con probabilidad de 0.95.

Además, supongamos que la probabilidad de estar infectado es de 0.017 (1.7%).

Ahora, suponga que una persona tuvo un resultado de prueba positivo.

1. ¿Cuál es la probabilidad de que esta persona esté infectada en efecto?

2. ¿Cuál es la probabilidad de que no esté infectada?

Primero vamos a ver cuales son los datos que tenemos

1. $p(cov = 1 | inf = 1) = 0.95$
2. $p(cov = 0 | inf = 0) = 0.95$
3. $p(inf = 1) = 0.017$

Pregunta 1

Queremos encontrar $p(\text{inf = 1 | cov = 1}) = ~ ?$

$$
p(\text{inf = 1 | cov = 1}) = \frac{p(cov = 1 | inf = 1) p(inf)}{p(cov)}
$$

Y para $p(cov)$ seria de la siguiente manera 

$$
p(cov) = p(cov | inf = 1)\cdot p(inf = 1) + p(cov | inf = 0) \cdot p(inf = 0) \\
p(cov) = 0.95 \cdot 0.017 + 0.05 \cdot 0.983 = 0.0653
$$


Ahora si sustituimos todo en nuestra formula

$$
p(\text{inf = 1 | cov = 1}) = \dfrac{0.95 \cdot 0.017}{0.0653} = 0.2473
$$

Si una persona tuvo un resultado de prueba positivo la probabilidad de que esta persona esté infectada en efecto es de **0.2473**.

Pregunta 2

Ahora queremos encontrar $p(\text{inf = 0 | cov = 1}) = ~ ?$

$$
p(\text{inf = 0 | cov = 1}) = 1 - p(\text{inf = 1 | cov = 1}) = 1 - 0.2473 = 0.7527
$$

Si una persona tuvo un resultado de prueba positivo la probabilidad de que NO esté infectada es de **0.7527**

(Ejercicio 1.3 Pattern Recognition and Machine Learning, by Christopher M. Bishop.) 

Supongamos que tenemos tres cajas: $r$ (roja), $a$ (azul), y $v$ (verde). Los contenidos de las cajas son:

- $r$: 3 manzanas, 4 naranjas, y 3 limas.
- $a$: 1 manzanas, 1 naranjas, y 0 limas.
- $v$: 3 manzanas, 3 naranjas, y 4 limas

Si la caja se elige aleatoriamente con probabilidades $p(r) = 0.2$, $p(b) = 0.2$, $p(g) = 0.6$, y una fruta se selecciona de la caja elegida, con igual probabilidad de seleccionar cualquier elemento de la caja, 

1. ¿Cuál es la probabilida de seleccionar una manzana?
2. Si observamos que la fruta seleccionada es una naranja, ¿Cuál es la probabilidad de que provenga de la caja verde?

Lo primero que vamos a hacer es saber cuales son todas las probabilidades 

1. Para la caja roja es la siguiente 
    * $p(r) = 0.2$
    * $p(m | r) = \frac{3}{10} = 0.3$
    * $p(n | r) = \frac{4}{10} = 0.4$
    * $p(l | r) = \frac{3}{10} = 0.3$

2. Para la caja azul es la siguiente
    * $p(a) = 0.2$
    * $p(m | a) = \frac{1}{2} = 0.5$
    * $p(n | a) = \frac{1}{2} = 0.5$
    * $p(l | a) = \frac{0}{2} = 0$

2. Para la caja verde es la siguiente
    * $p(v) = 0.6$
    * $p(m | v) = \frac{3}{10} = 0.3$
    * $p(n | v) = \frac{3}{10} = 0.3$
    * $p(l | v) = \frac{4}{10} = 0.4$ 

Pregunta 1

Queremos encontrar $p(m) =$ ?. Sabemso que 

$$
p(m) = \sum_x{p(m | x)p(x)}
$$

donde `x` es cada caja, entonces queda de la siguiente manera

$$
p(m) = p(m | r) \cdot p(r) + p(m | a) \cdot p(a) + p(m | v) \cdot p(v)
$$

$$
p(m) = 0.3 \cdot 0.2 + 0.5 \cdot 0.2 + 0.3 \cdot 0.6 = 0.34
$$

La probabilidad de seleccionar una manzana es de **0.34**

Pregunta 2

Queremos encontrar $p(v|n) =$ ? 

$$
p(v|n) = \dfrac{p(n|v) \cdot p(v)}{p(n)}
$$

vamos a buscar $p(n)$

$$
p(n) = p(n | r) \cdot p(r) + p(n | a) \cdot p(a) + p(n | v) \cdot p(v)
$$

$$
p(m) = 0.4 \cdot 0.2 + 0.5 \cdot 0.2 + 0.3 \cdot 0.6 = 0.36
$$

Sustituimos todo en la formula 

$$
p(v|n) = \dfrac{0.3 \cdot 0.6}{0.36} = 0.5
$$

Si la fruta seleccionada es naranja la probabilidad de que venga de la caja verde es de **0.50**

Leer la **sección 1.2.4. (The Gaussian distribution)** del libro Pattern Recognition and Machine Learning, by Christopher M. Bishop, las páginas 24 y 25 . A partir de ello:

1. ¿Cuáles son los parámetros que caracterizan una distribución Gaussiana?, ¿Qué significan?
   
2. Define una V.A. normal usando `scipy.stats.norm`, con media y varianza que prefieras:
   
   ```python
   x = stats.norm(loc=mu, scale=sigma)
   ```
   
   Luego, muestrea 10, 100, 1000, y 10000 veces esta V.A. usando el método `rvs`:
   
   ```python
   samples = x.rvs(n)
   ```
   
   Finalmente, toma el promedio simple de cada uno de los conjuntos de muestras:
   
   ```python
   samples.mean()
   ```
   
   ¿Qué puedes concluir del efecto de la cantidad de las muestras sobre este promedio?

Pregunta 1

Los parametros que caracterizan a una distribución Gaussiana son 2. La primera es $\mu$ que es la media y la otra es $\sigma^2$ que es la varianza.

La $\mu$ es el valor esperados de la distribución Gaussiana, que tiene sentido en que sea el promedio.

La $\sigma^2$ es la dispersion de los datos respecto a la media. 

Pregunta 2

````python
from scipy import stats

# Creamos la V.A.
mu = 2
sigma = 1
x = stats.norm(loc=mu, scale=sigma)

# Hacemos los sample
samples = {
    10: x.rvs(10),
    100: x.rvs(100),
    1000: x.rvs(1000),
    10000: x.rvs(10000),
}

for sam, mues in samples.items():
    print(f'Numero de sample {sam}. Su media es {mues.mean()}')
````

A la conclusión que llego despues de repetir el codigo varia veces, es que entre mas muestras hay mas se acerca al valor esperado o a $\mu$ de manera mas precisa. Porque en cada iteracion los valores en el de 10 y 100 samples cambiaba bastante, pero en el de mayor muestra las varaciones son mas pequeñas y cercanas a la media.

## 6.

Es común que haya cierta confusión en el significado del valor esperado. Un significado que le podemos dar a este valor, viene dado por la [ley de los grandes números](https://www.khanacademy.org/math/statistics-probability/random-variables-stats-library/expected-value-lib/v/law-of-large-numbers), que básicamente dice que:

$$
\mathbb{E}_p[f] \to \frac{1}{N} \sum_{i=1}^N f(x_i), \text{ cuando } N \to \infty
$$

Para convencerte de esto, en este ejercicio simularás la tirada de un dado justo:

1. Definimos la VA $X$ como el valor que muestra el dado cuando cae, es decir $\mathrm{Val}(X) = \{1, 2, 3, 4, 5, 6\}$.
2. Como el dado es justo $P(X=i)=\frac{1}{|\mathrm{Val}(X)|}=\frac{1}{6}$ para todo $i=1,\dots,6$.

¿Cuál es el valor esperado $E[X]$?

El valor esperado en una V.A. discreta es

$$
\mathbb{E}[X] = \sum_x p(x)x.
$$

Entonces el valor esperado es
$$
\mathbb{E}[X] = \dfrac{1}{6} \cdot 1 + \dfrac{1}{6} \cdot 2 + \dfrac{1}{6} \cdot 3 + \dfrac{1}{6} \cdot 4 + \dfrac{1}{6} \cdot 5 + \dfrac{1}{6} \cdot 6 = 3.5$$

El valor esperado segun el calculo es de **3.5**

Ahora veamos en codigo

````python
# Calcular el valor esperado teórico de la VA X
dado = stats.rv_discrete(
    name='dado',
    values=(
        [1,2,3,4,5,6], # xk
        [1/6,1/6,1/6,1/6,1/6,1/6] # pk
    )
)
# Valor esperado teórico
dado.expect()
````

Vemos que tambien nos da el resutlado de **3.5**

Ahora, viene el paso de simulación. 

1. Vas a simular por lo menos 1000 veces la tirada del dado, guardando a cada tirada el resultado obtenido.

2. Vas a calcular el valor promedio de los resultados *a cada tirada*. Es decir, si los resultados en las primeras $n$ tiradas son $x_1,\dots,x_n$, el valor promedio en la tirada $n$ es

   $$\bar{x}_n = \frac{x_1+\dots+x_n}{n}.$$

3. Finalmente, vas a graficar los valores promedio obtenidos a cada tirada (en el eje $y$) vs. el número de tiradas (en el eje $x$). Incluir el valor esperado teórico en la gráfica. ¿Qué observas?

*Ayuda: apoyarse del modulo random de numpy.*

````python
# Simular 1000 veces
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

n_samples = 1500
samples = dado.rvs(size=n_samples) # con stats
# samples = np.random.randint(1,7,(1500,1)) # con numpy
````

````python
# Calcular el valor promedio de los resultados a cada tirada
accu_sum = np.cumsum(samples)
average = accu_sum / np.arange(1,n_samples+1)
````

````python
sns.lineplot(average, label='Promedio sucesivo de muestra')
plt.axhline(y=dado.expect(), color='#F09D8B', label=r'$\mathbb{E}[X]$')
plt.xlabel('Numero de muestras')
plt.ylabel('Promedio Sucesivo')
plt.grid()
plt.legend()
plt.show()
````

Lo que puede observar corriendo varias muestras es que a mayor cantidad de datos, el promedio del resultado de los dados tiende al valor esperado. El cual fue antes calculado y es de **3.5**.
